# LinguaBridge: use AI in your own language, at a fair price

**Veritas AI Hackathon 2026 · AI + Languages track**

AI models charge by the *token*. The same message costs more tokens, and so more money and more of the model's memory, in some languages than in English. LinguaBridge measures that tax for Spanish, Hindi, Mandarin, Arabic and Kurmanji, and bridges questions through English when that gives a better answer.

**Before you start:** Runtime → Change runtime type → **T4 GPU** (needed for steps 4 to 6 only).
